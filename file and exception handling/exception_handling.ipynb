```python
# DICTINOARY KEY HANDLING
#Create a dictonary with students name and marks.
#ask user to enter a key Handel KeyError if key exists, print marks using else

try:
    students={
        "shoaib":85,
        "Hardhik":90,
        "manju":88,
        "Akhil": 77
    }

    name = input("Enter student name:")

    marks = students[name]

except KeyError:
    print("Error: Student not found in dictionary!")

else:
    print("marks of", name,"is",marks)
```

```python
units = int(input("Enter units consumed: "))

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = 100 * 5 + (units - 100) * 7
else:
    bill = 100 * 5 + 100 * 7 + (units - 200) * 10

print("Electricity Bill = ₹", bill)
```

```python
balance = 5000

while True:
    print("\n1.Check Balance")
    print("2.Deposit")
    print("3.Withdraw")
    print("4.Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        print("Balance =", balance)

    elif choice == 2:
        amount = float(input("Enter deposit amount: "))
        balance += amount
        print("Updated Balance =", balance)

    elif choice == 3:
        amount = float(input("Enter withdrawal amount: "))
        if amount <= balance:
            balance -= amount
            print("Remaining Balance =", balance)
        else:
            print("Insufficient Balance")

    elif choice == 4:
        print("Thank You")
        break

    else:
        print("Invalid Choice")
```
