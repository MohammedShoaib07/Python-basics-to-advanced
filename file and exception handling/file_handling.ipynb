```python
## WRITING INTO A FILE

with open("sample_Text.txt","w")as file:
    file.write("HELLO FROM VS CODE\n")
    file.write("FILE HANDLING IS WORKING\n")
print("file created successfully")
```

```python
## APPEND

with open("sample_Text.txt","a")as file:
    file.write("\nHELLO FROM 5th sem\n")
    file.write("MY NAME IS SHOAIB\n")
    file.write("FILE HANDLING IS WORKING")
print("file created successfully")
```

```python
##READING DATA

with open("Sample_Text.txt","r") as file:
    content=file.read()
    print(content)
```
