```python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.__price = price

    def get_price(self):
        return self.__price

    def set_price(self, price):
        if price >= 0:
            self.__price = price
        else:
            raise ValueError("Price cannot be negative")

    def display_details(self):
        print(f"Title: {self.title}")
        print(f"Author: {self.author}")
        print(f"Price: {self.get_price()}")


class Ebook(Book):
    def __init__(self, title, author, price, file_size):
        super().__init__(title, author, price)
        self.file_size = file_size

    def display_details(self):
        print(f"Title: {self.title}")
        print(f"Author: {self.author}")
        print(f"Price: {self.get_price()}")
        print(f"File Size: {self.file_size} MB")


book1 = Book("Python Basics", "shoaib", 500)
ebook1 = Ebook("Python OOP", "Talha", 300, 25)

book1.display_details()
print()
ebook1.display_details()
```

```python
class Vehicle:
    def __init__(self, brand, rent_per_day):
        self.brand = brand
        self.rent_per_day = rent_per_day

class Car(Vehicle):
    def calculate_rent(self, days):
        return self.rent_per_day * days
    
class Bike(Vehicle):
    def calculate_rent(self, days):
        return self.rent_per_day * days * 0.9

c = Car("Toyota", 1000)
b = Bike("Honda", 500)

print("Car Rent:", c.calculate_rent(3))
print("Bike Rent:", b.calculate_rent(3))
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show_details(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")


class Student(Person):
    def __init__(self, name, age, grade):
        super().__init__(name, age)
        self.grade = grade

    def show_details(self):
        super().show_details()
        print(f"Role: Student")
        print(f"Grade: {self.grade}")


class Teacher(Person):
    def __init__(self, name, age, subject):
        super().__init__(name, age)
        self.subject = subject

    def show_details(self):
        super().show_details()
        print(f"Role: Teacher")
        print(f"Subject: {self.subject}")


people = [
    Student("Ali", 16, "A"),
    Teacher("Ms. Sara", 35, "Math")
]

for person in people:
    person.show_details()
    print("-" * 20)
```
