```python
## CHECK FOR STRONG NUMBER 
num = int(input("enter number to check : "))

f=num//100
m=(num//10)%10
l=num%10
def fact_nums(f,m,l):
    if f==0 or f==1:
        return 1
    else:
        fact1 = 1
        for i in range(1, f+1):
            fact1 = fact1 * i

    if m==0 or m==1:
        return 1
    else:
        fact2 = 1
        for i in range(1, m+1):
            fact2 = fact2 * i

    if l==0 or l==1:
        return 1
    else:
        fact3 = 1
        for i in range(1, l+1):
            fact3 = fact3 * i

    sum=fact1+fact2+fact3
    return(sum)

if fact_nums(f,m,l) ==num:
    print("strong number")
else:
    print("not a strong number")


```

```python
##

employees = [
    (101, "shoaib", "AIML", 45000),
    (102, "talha", "MECH", 60000),
    (103, "afaaz", "CSE", 55000),
    (104, "sai", "ECE", 70000)
]

for ele in employees:
    print(f"ID : {ele[0]}\nNAME : {ele[1]}\nDEPT : {ele[2]}\nSalary : {ele[3]}\n")

highest = employees[0]

for emp in employees:
    if emp[3] > highest[3]:
        highest = emp

print("\nEmployee with Highest Salary:")
print("ID:", highest[0])
print("Name:", highest[1])
print("Department:", highest[2])
print("Salary:", highest[3])
```
