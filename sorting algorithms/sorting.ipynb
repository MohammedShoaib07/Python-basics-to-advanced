```python
arr = list(map(int, input("Enter the elements: ").split()))
n = len(arr)

for i in range(n - 1):
    min_index = i

    for j in range(i + 1, n):
        if arr[j] < arr[min_index]:
            min_index = j
            
    arr[i], arr[min_index] = arr[min_index], arr[i]

print("Sorted array:", arr)
```

```python
 #Insertion Sort

arr = list(map(int, input("Enter the elements: ").split()))
n = len(arr)

for i in range(1, n):
    key = arr[i]
    j = i - 1

    while j >= 0 and arr[j] > key:
        arr[j + 1] = arr[j]
        j -= 1

    arr[j + 1] = key

print("Sorted array:", arr)
```

```python
# Merge Sort

def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr) // 2

        left = arr[:mid]
        right = arr[mid:]

        # Recursively sort both halves
        merge_sort(left)
        merge_sort(right)

        i = j = k = 0

        # Merge the sorted halves
        while i < len(left) and j < len(right):
            if left[i] < right[j]:
                arr[k] = left[i]
                i += 1
            else:
                arr[k] = right[j]
                j += 1
            k += 1

        # Copy remaining elements of left half
        while i < len(left):
            arr[k] = left[i]
            i += 1
            k += 1

        # Copy remaining elements of right half
        while j < len(right):
            arr[k] = right[j]
            j += 1
            k += 1

# Main Program
arr = list(map(int, input("Enter the elements: ").split()))

merge_sort(arr)

print("Sorted array:", arr)
```

```python
n = int(input("Enter the number of elements: "))

arr = []
for i in range(n):
    arr.append(int(input(f"Enter element {i+1}: ")))

leaders = []
max_right = arr[-1]
leaders.append(max_right)

for i in range(n - 2, -1, -1):
    if arr[i] > max_right:
        leaders.append(arr[i])
        max_right = arr[i]

leaders.reverse()

print("Leaders in the array are:")
for leader in leaders:
    print(leader, end=" ")
```

```python
n = int(input("Enter the number of elements: "))

arr = []

print("Enter the elements:")
for i in range(n):
    arr.append(int(input()))

max_right = -1

for i in range(n - 1, -1, -1):
    current = arr[i]
    arr[i] = max_right
    if current > max_right:
        max_right = current

print("Array after replacement:")
print(arr)
```

```python
n = int(input("Enter the number of elements: "))

arr = []

print("Enter the elements:")
for i in range(n):
    arr.append(int(input()))

print("Duplicate elements are:")

duplicates = []

for i in range(n):
    if arr[i] not in duplicates:
        count = 0
        for j in range(n):
            if arr[i] == arr[j]:
                count += 1
        if count > 1:
            duplicates.append(arr[i])

for i in duplicates:
    print(i, end=" ")
```

```python
n = int(input("Enter the number of elements: "))

arr = []

print("Enter the elements:")
for i in range(n):
    arr.append(int(input()))

positive = []
negative = []

for i in arr:
    if i >= 0:
        positive.append(i)
    else:
        negative.append(i)

result = []

i = 0
j = 0

while i < len(positive) and j < len(negative):
    result.append(positive[i])
    result.append(negative[j])
    i += 1
    j += 1

while i < len(positive):
    result.append(positive[i])
    i += 1

while j < len(negative):
    result.append(negative[j])
    j += 1

print("Rearranged array:")
print(result)
```

```python
n = int(input("Enter the number of elements: "))

arr = []

print("Enter the elements:")
for i in range(n):
    arr.append(int(input()))

max_len = 1
curr_len = 1

start = 0
end = 0
temp_start = 0

for i in range(1, n):
    if arr[i] > arr[i - 1]:
        curr_len += 1
    else:
        if curr_len > max_len:
            max_len = curr_len
            start = temp_start
            end = i - 1
        curr_len = 1
        temp_start = i

if curr_len > max_len:
    max_len = curr_len
    start = temp_start
    end = n - 1

print("Longest Increasing Subarray:")
print(arr[start:end + 1])

print("Length:")
print(max_len)

print("Starting Index:")
print(start)

print("Ending Index:")
print(end)
```
