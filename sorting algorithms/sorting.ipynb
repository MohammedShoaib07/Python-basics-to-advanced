```python
a=[3,5,1,2,8,4]
n=len(a)

for i in range(n-1):
    for j in range(n-i-1):
        if a[j]>a[j+1]:
           a[j], a[j+1]=a[j+1],a[j]

print(a) 
```

```python
arr = list(map(int, input("Enter the elements: ").split()))
n = len(arr)

for i in range(n - 1):
    min_index = i

    for j in range(i + 1, n):
        if arr[j] < arr[min_index]:
            min_index = j
            
    arr[i], arr[min_index] = arr[min_index], arr[i]

print("Sorted array:", arr)
```
